# Test data drift

```python
reference = pd.read_csv("application_train.csv", sep = ",", low_memory=False).drop(columns = "TARGET")
reference.shape
```

```python
current = pd.read_csv("application_test.csv", sep = ",", low_memory=False)
current.shape
```

```python
columns = reference.select_dtypes(include=['int64', 'float64']).columns
columns.shape

# Génération des métriques
metrics = [ValueDrift(column=col) for col in columns]

# Création et exécution du rapport
report = Report(metrics=metrics)
report.run(reference_data=reference, current_data=current)
#report.show(mode='inline')
```
